```python
from ultralytics import YOLO

# 1. 학습이 완료된 가중치 파일 불러오기
model = YOLO("/workspace/welding_vision_project/sub_assembly_v1/weights/best.pt")

# 2. 검증(Validation) 실행 및 메트릭 확인
# (train.py에서 사용했던 data.yaml 경로를 그대로 지정해주어야 합니다)
metrics = model.val(data="/workspace/dataset/data.yaml")

# 3. 주요 정확도 지표 출력
print(f"mAP50-95: {metrics.box.map}")  # 가장 종합적인 정확도 지표 (0 ~ 1 사이 값)
print(f"mAP50: {metrics.box.map50}")  # IoU 0.5 기준 정확도
print(f"mAP75: {metrics.box.map75}")  # IoU 0.75 기준 정확도
```
